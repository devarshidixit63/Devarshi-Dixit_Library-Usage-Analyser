## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated columns

# 1. Calculate return rate percentage
df["Return_Rate"] = (
    df["Books_Returned"] / df["Books_Issued"]
) * 100

# 2. Calculate difference between issued and returned books
df["Books_Difference"] = (
    df["Books_Issued"] - df["Books_Returned"]
)

print("Dataset with New Columns:")
display(df.head())


# TODO: Filter records using a condition

# Students with more than 2 late returns
late_students = df[df["Late_Returns"] > 2]

print("\nStudents with More Than 2 Late Returns:")
display(late_students)


# TODO: Sort records and display top/bottom results

# Top 5 students with most late returns
top_students = df.sort_values(
    by="Late_Returns", ascending=False
)

print("\nTop 5 Students with Most Late Returns:")
display(top_students.head())

# Bottom 5 students with least late returns
print("\nBottom 5 Students with Least Late Returns:")
display(top_students.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate return rate; identify students with most late returns; summarise usage.

Write your solution below and explain what the result means.


```python
# 1. Calculate return rate for each student
df["Return_Rate"] = (df["Books_Returned"] / df["Books_Issued"]) * 100

print("Student Return Rates:")
display(df[["Name", "Books_Issued", "Books_Returned", "Return_Rate"]])


# 2. Identify students with most late returns
max_late = df["Late_Returns"].max()

students = df[df["Late_Returns"] == max_late]

print("\nStudents with Most Late Returns:")
display(students[["Name", "Late_Returns"]])


# 3. Summarise library usage
total_issued = df["Books_Issued"].sum()
total_returned = df["Books_Returned"].sum()
total_late = df["Late_Returns"].sum()

print("\nLibrary Usage Summary:")
print("Total Books Issued:", total_issued)
print("Total Books Returned:", total_returned)
print("Total Late Returns:", total_late)


# 4. Calculate overall return rate
overall_rate = (total_returned / total_issued) * 100

print("Overall Return Rate:", round(overall_rate, 2), "%")


# 5. Save final analysis to CSV
df.to_csv("Student_11_Final_Analysis.csv", index=False)

print("\nFinal analysis saved successfully!")

```
